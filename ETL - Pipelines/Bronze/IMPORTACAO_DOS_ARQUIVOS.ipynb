```python
dbutils.widgets.text("CAMINHO_ARQUIVO", "")
```

```python
path = dbutils.widgets.get("CAMINHO_ARQUIVO")



for file_info in dbutils.fs.ls(path):
    if file_info.path.endswith('.csv'):

        if file_info.name == "Canais.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("append").saveAsTable("vendas_delivery.bronze.CANAIS")
        
        elif file_info.name == "Centro_de_Distruibuicao.csv":
            #Tratativa necessária para não quebrar os caracteres latinos dos estados
            df = spark.read.option("encoding", "ISO-8859-1").csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.CENTRO_DE_DISTRIBUICAO")

        elif file_info.name == "Entegador.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.ENTREGADOR")

        elif file_info.name == "Entregas.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.ENTREGAS")

        elif file_info.name == "lojas.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.LOJAS")

        elif file_info.name == "pagamentos.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.PAGAMENTOS")

        elif file_info.name == "Pedidos.csv":
            df = spark.read.csv(file_info.path, header=True, inferSchema=True)
            df.write.format("delta").mode("overwrite").saveAsTable("vendas_delivery.bronze.PEDIDOS")
            
        
        else:
            print("Erro ao ler o arquivo: " + file_info.name)
        
```
