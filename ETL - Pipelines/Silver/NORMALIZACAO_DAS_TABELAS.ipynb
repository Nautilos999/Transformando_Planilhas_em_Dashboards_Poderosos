```python
from pyspark.sql.functions import col, upper

#Garantir que as colunas de descrição estejam sempre maiúsculas
df_canal = spark.sql("select channel_id, channel_name, channel_type from vendas_delivery.bronze.canais")
df_canal = df_canal.withColumn("channel_name", upper(col("channel_name")))
df_canal = df_canal.withColumn("channel_type", upper(col("channel_type")))

df_canal.write.mode("overwrite").saveAsTable("vendas_delivery.silver.canais")
```

```python

#Tratativa para trzer somente as colunas necessárias
df_centro = spark.sql("select hub_id, hub_name, hub_city, hub_state from vendas_delivery.bronze.centro_de_distribuicao")
df_centro = df_centro.withColumn("hub_name", upper(col("hub_name")))
df_centro = df_centro.withColumn("hub_city", upper(col("hub_city")))
df_centro = df_centro.withColumn("hub_state", upper(col("hub_state")))

df_centro.write.mode("overwrite").saveAsTable("vendas_delivery.silver.centro_de_distribuicao")
```

```python

#Garantir que as colunas de descrição estejam sempre maiúsculas
df_entregador = spark.sql("select driver_id, driver_modal, driver_type from vendas_delivery.bronze.entregador")
df_entregador = df_entregador.withColumn("driver_modal", upper(col("driver_modal")))
df_entregador = df_entregador.withColumn("driver_type", upper(col("driver_type")))

df_entregador.write.mode("overwrite").saveAsTable("vendas_delivery.silver.entregador")
```

```python

#Garante que a coluna delivery_status esteja sempre maiúscula
#Trata os valores de entregadores NULL para -1 
#Converte os valores de distancia em metros para km
df_entregas = spark.sql("""select delivery_id, delivery_order_id, 
                      case when driver_id is null then -1 else driver_id end as driver_id,
                      cast((delivery_distance_meters / 1000) as decimal(10,2)) as delivery_distance_km, 
                      delivery_status from vendas_delivery.bronze.entregas""")
df_entregas = df_entregas.withColumn("delivery_status", upper(col("delivery_status")))

df_entregas.write.mode("overwrite").saveAsTable("vendas_delivery.silver.entregas")
```
